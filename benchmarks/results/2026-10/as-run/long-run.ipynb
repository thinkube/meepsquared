```python
PARAMETERS = globals().get('PARAMETERS', {})
CMD = PARAMETERS['cmd']          # the benchmark command, as a list
CWD = PARAMETERS['cwd']
```

```python
import subprocess, sys, time
print(time.strftime('%H:%M:%S'), ' '.join(CMD), flush=True)
p = subprocess.run([sys.executable, '-u', *CMD], cwd=CWD, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, timeout=5 * 3600)
lines = p.stdout.splitlines()
print('\n'.join(l for l in lines if 'RESULT' in l or l.startswith('llebre') or 'Error' in l or 'Traceback' in l or 'rror:' in l)[-8000:])
print(time.strftime('%H:%M:%S'), 'EXIT', p.returncode)
if p.returncode != 0:
    print('\n'.join(lines[-60:]))
    raise SystemExit(p.returncode)
```
